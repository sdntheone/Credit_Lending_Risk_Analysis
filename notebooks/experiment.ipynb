```python
def treat_columns(df, columns_treated):
    import numpy as np
    
    time_cols = ['time_since_recent_payment', 'time_since_recent_enq']
    ratio_cols = ['pct_currentBal_all_TL']
    count_cols = [col for col in columns_treated if col not in time_cols + ratio_cols]
    
    df[columns_treated] = df[columns_treated].replace(-99999, np.nan)
    
    for col in time_cols:
        df[col + "_missing"] = df[col].isna().astype(int)
        df[col] = df[col].fillna(df[col].max() + 1)
    
    df[count_cols] = df[count_cols].fillna(0)
    
    for col in ratio_cols:
        df[col].fillna(df[col].median(), inplace=True)
    
    return df
```

```python
# Chi-square test
from scipy.stats import chi2_contingency
for i in ['MARITALSTATUS', 'EDUCATION', 'GENDER', 'last_prod_enq2', 'first_prod_enq2']:
    chi2, pval, _, _ = chi2_contingency(pd.crosstab(df[i], df['Approved_Flag']))
    print(i, '---', pval)
```

since all the categorical features have p-value less than 0.05, we will accept all the variable


```python
# VIF for numerical columns
numeric_columns = []
for i in df.columns:
    if df[i].dtype != 'object' and i not in ['PROSPECTID','Approved_Flag']:
        numeric_columns.append(i)
```

```python
len(numeric_columns)
```

```python
# Selecting numerical features
numeric_columns = df.select_dtypes(include=[np.number]).columns.tolist()
vif_data = df[numeric_columns].copy()

# Set VIF threshold
vif_threshold = 6
columns_to_be_kept = []
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
while True:
    vif_values = [variance_inflation_factor(vif_data.values, i) for i in range(vif_data.shape[1])]
    vif_df = pd.DataFrame({"Feature": vif_data.columns, "VIF": vif_values})
    
    # Print VIF values
    print(vif_df.sort_values(by="VIF", ascending=False))
    
    max_vif = vif_df["VIF"].max()
    
    if max_vif < vif_threshold:
        break  # Stop if all VIFs are below threshold
    
    # Drop feature with highest VIF
    drop_feature = vif_df.loc[vif_df["VIF"] == max_vif, "Feature"].values[0]
    print(f"Dropping {drop_feature} with VIF: {max_vif}")
    vif_data.drop(columns=[drop_feature], inplace=True)

# Store final features with low VIF
columns_to_be_kept = vif_data.columns.tolist()

print("\nFinal Features with Low VIF:\n", columns_to_be_kept)
```

```python
len(columns_to_be_kept)
```

```python
# check Anova for columns_to_be_kept 

from scipy.stats import f_oneway

columns_to_be_kept_numerical = []

for i in columns_to_be_kept:
    a = list(df[i])  
    b = list(df['Approved_Flag'])  
    
    group_P1 = [value for value, group in zip(a, b) if group == 'P1']
    group_P2 = [value for value, group in zip(a, b) if group == 'P2']
    group_P3 = [value for value, group in zip(a, b) if group == 'P3']
    group_P4 = [value for value, group in zip(a, b) if group == 'P4']


    f_statistic, p_value = f_oneway(group_P1, group_P2, group_P3, group_P4)

    if p_value <= 0.05:
        columns_to_be_kept_numerical.append(i)
```

## Feature engineering

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
from imblearn.over_sampling import SMOTENC
```

```python
test_size = 0.2   # load from params.yaml
random_state = 42
```

```python
df['Approved_Flag'].value_counts()
```

```python
# 1. Encode target
label_encoder = LabelEncoder()
df['Approved_Flag'] = label_encoder.fit_transform(df['Approved_Flag'])

X = df.drop(columns=['Approved_Flag'])
y = df['Approved_Flag']
```

```python
# 2. Split FIRST
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=test_size, stratify=y, random_state=random_state
)
```

```python
categorical_cols = ['EDUCATION', 'GENDER', 'last_prod_enq2', 'first_prod_enq2', 'MARITALSTATUS']
```

```python
encoder = OrdinalEncoder()

X_train[categorical_cols] = encoder.fit_transform(X_train[categorical_cols])
X_test[categorical_cols] = encoder.transform(X_test[categorical_cols])
```

```python
categorical_indices = [X_train.columns.get_loc(col) for col in categorical_cols]
```

```python
# 6. SMOTENC (ONLY on train)
smote_nc = SMOTENC(
    categorical_features=categorical_indices,
    random_state=random_state
)
```

```python
X_train_resampled, y_train_resampled = smote_nc.fit_resample(X_train, y_train)
```

```python
train_final_df = pd.concat(
    [pd.DataFrame(X_train_resampled, columns=X_train.columns),
     pd.Series(y_train_resampled, name='Approved_Flag')],
    axis=1
)
```

```python
train_final_df.shape
```

```python
train_final_df['Approved_Flag'].value_counts()
```

```python
test_final_df = pd.concat(
    [X_test.reset_index(drop=True),
     y_test.reset_index(drop=True)],
    axis=1
)
```

```python
test_final_df.shape
```

```python
test_final_df['Approved_Flag'].value_counts()
```

```python
X_train = train_final_df.drop('Approved_Flag', axis=1)
y_train = train_final_df['Approved_Flag']

X_test = test_final_df.drop('Approved_Flag', axis=1)
y_test = test_final_df['Approved_Flag']

class_names = ['p1', 'p2', 'p3', 'p4']
```

```python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, f1_score

def evaluate_model(name, model, X_train, y_train, X_test, y_test):
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    accuracy = accuracy_score(y_test, y_pred)
    precision, recall, f1_vals, _ = precision_recall_fscore_support(y_test, y_pred)
    
    f1_weighted = f1_score(y_test, y_pred, average='weighted')
    f1_macro = f1_score(y_test, y_pred, average='macro')
    
    print(f"\n===== {name} =====")
    print(f"Accuracy: {accuracy}")
    print(f"F1 Weighted: {f1_weighted}")
    print(f"F1 Macro: {f1_macro}\n")
    
    for i, v in enumerate(class_names):
        print(f"Class {v}:")
        print(f"Precision: {precision[i]}")
        print(f"Recall: {recall[i]}")
        print(f"F1 Score: {f1_vals[i]}")
        print()
    
    return {
        "model": model,
        "accuracy": accuracy,
        "f1_weighted": f1_weighted,
        "f1_macro": f1_macro,
        "class_f1": f1_vals
    }
```

```python
from sklearn.ensemble import RandomForestClassifier

baseline_result = evaluate_model(
    "Random Forest (Baseline)",
    RandomForestClassifier(random_state=42),
    X_train, y_train, X_test, y_test
)
```

```python
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier

models = {
    "XGBoost": XGBClassifier(objective='multi:softprob', eval_metric='mlogloss'),
    "CatBoost": CatBoostClassifier(verbose=0),
    "LightGBM": LGBMClassifier()
}

results = []

for name, model in models.items():
    res = evaluate_model(name, model, X_train, y_train, X_test, y_test)
    results.append(res)
```

```python
model_names = list(models.keys())

for i, res in enumerate(results):
    res['name'] = model_names[i]
```

```python
best_result = max(
    results,
    key=lambda x: (x['f1_macro'], min(x['class_f1']))
)

print("\n===== BEST MODEL =====")
print("Model Name:", best_result['name'])   # ✅ ADD THIS
print("F1 Weighted:", best_result['f1_weighted'])
print("F1 Macro:", best_result['f1_macro'])
```

```python
fast_model = CatBoostClassifier(
    iterations=300,          
    depth=5,                 
    learning_rate=0.05,     
    l2_leaf_reg=5,           
    class_weights=[1, 1, 1.6, 1],
    verbose=0
)

fast_model.fit(X_train, y_train)
```

```python
evaluate_model(
    "Final catboost",
    fast_model,
    X_train, y_train, X_test, y_test
)
```
